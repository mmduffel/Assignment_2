**Part A: Missing Value Treatment & Part B: Removing Units**

```python
#Check for missing values in each column
df.isnull().sum()
```

```python
#Calculate the percentage of missing values in New_Price
(df['New_Price'].isnull().sum() / len(df)) * 100
```

```python
#Drop unnecessary columns
#Unnamed: 0 is an index column and New_Price has approximately 86% missing values
df = df.drop(columns=['Unnamed: 0', 'New_Price'])
```

```python
#Display the first 5 rows after dropping the columns
df.head()
```

```python
#Remove measurement units and convert columns to numeric values
df['Mileage'] = pd.to_numeric(df['Mileage'].str.split().str[0], errors='coerce')
df['Engine'] = pd.to_numeric(df['Engine'].str.split().str[0], errors='coerce')
df['Power'] = pd.to_numeric(df['Power'].str.split().str[0], errors='coerce')
```

```python
#Check the data types of selected columns
df[['Mileage', 'Engine', 'Power']].dtypes
```

```python
#Check missing values of selected columns
df[['Mileage', 'Engine', 'Power', 'Seats']].isnull().sum()
```

```python
#Fill missing numerical values with the median
df['Mileage'] = df['Mileage'].fillna(df['Mileage'].median())
df['Engine'] = df['Engine'].fillna(df['Engine'].median())
df['Power'] = df['Power'].fillna(df['Power'].median())

#Fill missing Seats values with the mode
df['Seats'] = df['Seats'].fillna(df['Seats'].mode()[0])
```

```python
#Check that all missing values have been handled
df.isnull().sum()
```

**Part A (explanation): Missing Value Treatment**

The dataset was checked for missing values. Mileage had 2 missing values, Engine had 36, Power had 36, Seats had 38, and New_Price had 5,032 missing values. The New_Price column was dropped because approximately 86.06% of its values were missing. Filling in that many missing values could affect the accuracy of the analysis and potentially introduce bias. Dropping all rows with missing New_Price values would also result in losing a large portion of the dataset.

The missing values in Mileage, Engine, and Power were replaced with their respective medians because the median is less affected by extreme values. Missing values in Seats were replaced with the mode because the number of seats is a discrete variable, making the most frequently occurring value a reasonable replacement.

The Unnamed: 0 column was also removed because it was an index column and did not provide meaningful information about the vehicles. After completing the cleaning and imputation process, the dataset was checked again, and no missing values remained.

**Part B (explanation): Removing Units**

The measurement units were removed from the Mileage, Engine, and Power columns so that only numerical values remained. For example, values such as 19.67 kmpl, 1582 CC, and 126.2 bhp were converted to 19.67, 1582, and 126.2, respectively. The columns were then converted to numerical data types so they could be used for further analysis and modeling.

New_Price was not included in this transformation because it was removed during the missing-value treatment in Part A. Approximately 86.06% of the values in New_Price were missing, so keeping this column and filling in the missing values was not considered appropriate.

**Part C: One-Hot Encoding**

```python
#Display the unique categories in Fuel_Type and Transmission
print(df['Fuel_Type'].unique())
print(df['Transmission'].unique())
```

```python
#Convert Fuel_Type and Transmission into one-hot encoded numerical variables
df = pd.get_dummies(df, columns=['Fuel_Type', 'Transmission'], dtype=int)
```

```python
#Display the first 5 rows to verify the one-hot encoded columns
df.head()
```

**Part C (explanation): One-Hot Encoding**

The categorical variables Fuel_Type and Transmission were converted into numerical variables using one-hot encoding. This process created separate binary columns for each category, where 1 indicates that the vehicle belongs to that category and 0 indicates that it does not. Fuel_Type was converted into Fuel_Type_Diesel, Fuel_Type_Electric, and Fuel_Type_Petrol, while Transmission was converted into Transmission_Automatic and Transmission_Manual. One-hot encoding prevents the categories from being assigned an artificial numerical order or ranking and allows the variables to be used in further analysis and modeling.

**Part D: Feature Engineering**

```python
#Create a new Car_Age feature
df['Car_Age'] = 2026 - df['Year']
```

```python
#Display Year and Car_Age to verify the new feature
df[['Year', 'Car_Age']].head()
```

```python
#Display the first 5 rows to verify new column Car_Age
df.head()
```

**Part D (explanation): Feature Engineering**

A new feature called Car_Age was created by subtracting the vehicle's Year of manufacture from the current year, 2026. This feature represents the approximate age of each vehicle in years. Car age may be useful when analyzing or predicting used-car prices because a vehicle's value can change as it gets older.

**Part E: Data Manipulation Operations**

```python
#Select specific columns from the dataset
selected_cars = df[['Name', 'Location', 'Year', 'Car_Age', 'Price']]

#Display the first 5 rows
selected_cars.head()
```

```python
#Filter the dataset to include cars priced above 10 lakhs
filtered_cars = df[df['Price'] > 10]

#Display the first 5 filtered records
filtered_cars.head()
```

```python
#Rename Kilometers_Driven to Odometer_Km
df = df.rename(columns={'Kilometers_Driven': 'Odometer_Km'})

#Display the updated column names
df.columns
```

```python
#Create a new feature for the average kilometers driven per year
#Replace a car age of 0 with 1 to prevent division by zero
df['Km_Per_Year'] = df['Odometer_Km'] / df['Car_Age'].replace(0, 1)

#Display the new feature
df[['Name', 'Odometer_Km', 'Car_Age', 'Km_Per_Year']].head()
```

```python
#Arrange the dataset from highest to lowest Price
arranged_cars = df.sort_values(by='Price', ascending=False)

#Display the 5 most expensive cars
arranged_cars[['Name', 'Year', 'Price']].head()
```

```python
#Group the dataset by Location and calculate summary statistics
location_summary = (
    df.groupby('Location')
      .agg(
          Number_of_Cars=('Price', 'count'),
          Average_Price=('Price', 'mean'),
          Median_Price=('Price', 'median'),
          Average_Mileage=('Mileage', 'mean')
      )
      .reset_index()
)

#Round the numerical results to 2 decimal places
location_summary = location_summary.round(2)

#Display the summary
location_summary
```

**Part E (explanation): Data Manipulation Operations**

Several data manipulation operations were performed using Python and pandas. The select operation was used to create a subset of the dataset containing the vehicle name, location, year, age, and price. The filter operation was used to identify vehicles priced above 10 lakhs. The Kilometers_Driven column was renamed Odometer_Km to provide a shorter and clearer description.

A mutate-equivalent operation was used to create Km_Per_Year, which estimates the average number of kilometers driven per year based on the vehicle's odometer reading and age. For vehicles with a Car_Age of 0, the value was temporarily treated as 1 in this calculation to prevent division by zero. The dataset was then arranged by price in descending order to identify the most expensive vehicles. Finally, the data was grouped by location and summarized by calculating the number of vehicles, average price, median price, and average mileage for each location.
